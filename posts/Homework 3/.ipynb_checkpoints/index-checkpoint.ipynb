## 3. Global Function Implementation

In the message bank, we need to access a database to store all the user inputs. We need a function that establishes a connection to the message database by reusing an existing connection or creating a new one if none exists. The `get_message_db` achieves the above purpose. The function uses the `sqlite3.connect` method to connect to the database (creating it if it doesn't exist). It checks if a table named messages exists in the database, and if not, it creates one with two columns: `handle` for storing user names and `message` for storing the actual text data. Finally, the function returns the connection.

````python
def get_message_db():
    """
    Checks whether a global variable `message_db` exists and is connected.
    If exists, function returns the database
    If not, connect to "messages_db.sqlite" database
            creates a `messages` table if it doesn't exist, 
            and returns the connection database.
    """
    global message_db
    # Return the database connection that already exists
    #if message_db:
        #return message_db
    #else:
    message_db = sqlite3.connect("messages_db.sqlite", check_same_thread = False)
    # SQL command to create the table if it doesn't exist
    cmd = 'CREATE TABLE IF NOT EXISTS messages (handle TEXT, message TEXT)'
    cursor = message_db.cursor()
    cursor.execute(cmd)
    return message_db
````

## 4. Functions for User Submission

After setting up a way to access the database, we now implement a function to store user input in the database. The function is defined as `insert_message`, which takes two parameters: `handle` for user name and `message` for the user's message. The function first initiates a connection to the message database. Then it uses an SQL query to insert the message and user name into the corresponding columns of the data table. Next, the changes are saved with `db.commit()`, which is crucial for ensuring the insertion is successful. Finally, the function closes the database connection.

````python
def insert_message(handle, message):
    """
    Inserting a user message into the database of messages
    Args:
        handle, string, the user name
        message, string, the message for insertion
    """
    # Get the database connection
    db = get_message_db() 
    cursor = db.cursor()
        
    # Insert input to database using SQL command
    cmd = f"""INSERT INTO messages (handle, message) VALUES ('{handle}', '{message}')"""
    cursor.execute(cmd)
    
    # Save the insertion
    db.commit()
    
    # Close the database connection
    db.close()
````

We have now reached the step of interacting with user input. We will begin by writing a callback function. The first output is feedback information that we provide to the user. If the user does not input either a message or a handle when submitting, the feedback will prompt the user to complete the input. If the user's submission fails, the feedback will indicate the error and tell the user to try again. In both of these scenarios, the last two outputs (the message and handle input boxes) will keep the user input. However, if the user successfully submits both the message and the handle, the feedback will express gratitude, and the input boxes will be cleared. The state in this context refers to the values that are retained between function calls. It allows the function to remember user input and react accordingly. The callback function will be triggered when the user clicks the submit button, meaning that the function will execute once the user submits their input. The `prevent_initial_call=True` parameter is used to prevent the callback function from executing when the app is first loaded, which ensures that the feedback or any other logic only occurs after user interaction.

Following this logic, we create a submit function that takes three parameters: "handle", "message", and "n_clicks". The function will only work when the submit button is activated with code `if n_clicks > 0`. It first checks if the user has input both the message and the handle. If both are provided, the function attempts to insert the message into the message database using the `insert_message` function we implemented earlier. Upon successful submission, the input fields are cleared. If there is an error during submission, the function will include the error message in the feedback.

An important thing to keep in mind is to maintain the same order throughout! You need to ensure that the state and input order in the callback function matches the order of parameters defined in the submit function. Additionally, the output order in the callback function should be consistent with the order of the values returned by the submit function. Keeping this order intact ensures that the inputs, states, and outputs are correctly mapped and processed, preventing any potential errors or mismatches in the data.

````python
@callback(
    Output(component_id = 'feedback', component_property = 'children'),
    Output(component_id = 'name_input', component_property = 'value'),
    Output(component_id = 'message_input', component_property = 'value'),
    State(component_id = 'name_input', component_property = 'value'),
    State(component_id = 'message_input', component_property = 'value'),
    Input(component_id = 'submit_button', component_property = 'n_clicks'),
    prevent_initial_call=True
)

def submit(handle, message, n_clicks):
    """
    Callback function to submit a message and display a feedback regarding submission
    Cleans the handle and message input if successfully submit the message
    Args:
        handle, string, the user name
        message, string, the message from user input
        n_clicks, integer, number of clicks on the submit button
    Returns:
        feedback, string, the feedback for user input
        clear_name, string, empty string that clear name input
        clear_message, string, empty string that clear message input
    """
    if n_clicks > 0:
        # Initialize feedback message
        feedback = ""
        clear_name = handle
        clear_message = message
    
        # Check if the user input both message and name
        if not message or not handle:
            feedback = "Please recheck both your name and message have been entered"
            return feedback, clear_name, clear_message
    
        # Try insert the message to the message database
        try:
            insert_message(handle, message)
            feedback = "Thank you for sharing your wisdom!"
            clear_name = ""
            clear_message = ""
            return feedback, clear_name, clear_message
        # Display error if submission is not successful
        except Exception as e:
            feedback = f"Oops... {e} Please try again..."
            return feedback, clear_name, clear_message
````

## 5. Functions for User View

In the review section, we will display random messages along with the corresponding handle. We begin by creating a function that accesses the database and retrieves the necessary messages.

n the `random_messages` function, we will use the `get_message_db` function again to connect to the message database. Next, we write an SQL query to randomly extract up to n rows from the database. After retrieving the messages, we close the database connection and make small modifications to the format of the messages and handles. We will store each message followed by its corresponding handle as individual list items and return the list. This approach allows us to easily display each message followed by a new line with the handle, ensuring clarity in the presentation.

````python
def random_messages(n):
    """
    Retrieve n random messages from the database.
    Args:
        n, integer, the number of random messages
    Returns:
        A list of tuples (handle, message)
    """
    # Get the database connection
    db = get_message_db()
    cursor = db.cursor()
    
    # Extract maximum n messages from the database with SQL command
    cmd = 'SELECT handle, message FROM messages ORDER BY RANDOM() LIMIT ?'
    cursor.execute(cmd, (n,))
    messages = cursor.fetchall()
    
    # Close the database connection
    db.close()
    
    # Initialize a list of formatted_messages
    formatted_messages = []
    
    # Define the format of the message display
    for handle, message in messages:
        formatted_messages.append(message)
        formatted_messages.append(f"-- {handle}")

    return formatted_messages
````

Lastly, we define a callback function that displays at most 5 randomly selected messages from the database. The function is triggered when the user clicks the update button. The `view` function takes n_clicks as its parameter which represents the number of times the submit button has been clicked. If the button has been clicked, it fetches 5 random messages from the database by calling the `random_messages` function with parameter integer of 5. After fetching the messages, uses a `for` loop to display the messages retrieved in the output component - 'random_message' text section. 

````python
@app.callback(
    Output(component_id = 'random_messages', component_property = 'children'),
    Input(component_id = 'update_button', component_property = 'n_clicks')
)

def view(n_clicks):
    """
    Callback function that displays at most 5 randomly selected messages from database
    Args:
        n_clicks, integer, number of clicks on the submit button
    Returns:
        html.Div, list of strings, random selected messages from database 
    """
    if n_clicks:
        # Fetch 5 random messages
        messages = random_messages(5)
        
        # Initialize a list of mssages
        message_list = []
        # Append the messages into the list
        for message in messages:
            message_list.append(message)
            message_list.append(html.Br()) 
        
        return message_list
````
